Find Largest and Smallest of numbers

```python
a = int(input("Enter First Number:"))
b = int(input("Enter Second Number:"))
c = int(input("Enter Third Number:"))
if a>b and a>c:
  print("a is greater")
elif b>a and b>c:
  print("b is greater")
else:
  print("c is greater")
```

Calculate Factorial

```python
num = int(input("Enter a num:"))
fact = 1

for i in range(num,0,-1):
    fact = fact * i
print(f"Factorial Of A Number: {fact}")
```

Check Prime Number

```python
num = int(input("Enter a number:"))
if num % num == 0:
    print(num," is prime number")
else:
    print(num," is not prime number")
```

Reverse a number

```python
num = 567
a = num // 100
b = num % 100
c = b % 10
b = b // 10
Reverse = c*100 + b*10 + a
print(f"Reverse Number = {Reverse} ")
```

Generate Fibonacci Series

```python
num =int(input("Enter a number:"))
first_num = 0
second_num = 1
next_num =  first_num + second_num
print(first_num)
print(second_num)
for i in range(num):
    next_num = next_num + i
    print(next_num)
```

Generate Palindrome

```python
num = int(input("Enter a number to check palindrome:"))
a = num // 100
b = num % 100
c = b % 10
b = b // 10
Reverse = c*100 + b*10 + a
print(f"The Given Number is:{num}")
print(f"Reverse Number = {Reverse} ")
if num == Reverse:
    print(f"{num} is a palindrome")
else:
    print(f"{num} is not a palindrome")
```
