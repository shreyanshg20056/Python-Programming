Fibonacci Series By Recursion

```python
def fibonacci(n):
  f1 = 0
  f2 = 1
  next = f1 + f2
  f1 = f2
  n += 1
  f2 += 1
  if n == 0 or n == 1:
    return 0
  else:
    print(next)
    fibonacci(n+1)

x = int(input("Enter N term:"))
fibonacci(x)
```
