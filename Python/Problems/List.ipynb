Add the number 99 to the end of a list, then insert 0 at the beginning.

```python
n = [1,3,6,8,4]
n.insert(5,99)
n.insert(0,0)
print(n)
```

Remove a given element from a list (for example, remove 15 from nums).

```python
n = [1,2,5,6,3,15]
n.remove(15)
print(n)
```

Find the length of a list without using len().

```python
n = [1,2,4,5,6]
count = 0
for i in n:
  if i != 0:
    count += 1

print(count)
```

Find the sum of all elements in a list

```python
n = [1,2,46,8,9]
s = 0
for i in n:
  s = s + i

print(s) 
```
