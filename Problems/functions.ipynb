Print Number from 1-5 using for loop

```python
for i in range(1,6):
  print(i)
```

Print Squares Of Numbers From 1-5 using for loop

```python
for i in range(1,6):
  print(i*i)
```

Print Even Numbers From 1-10 using for loop

```python
for i in range(1,11):
  if i % 2 == 0:
    print(i)
```

Calculate Sum Of Numbers From 1-10 using for loop

```python
sum = 0

for i in range(1,11):
  sum = sum + i
print("Sum:",sum)
```

Reverse A Word

```python
S = "Python"
print(S[ : :-1])
```

Count vowel in a string

```python
S = "Education"
count = 0

for i in S:
  if i == 'a':
    count += 1
  elif i == 'e':
    count += 1
  elif i == 'i':
    count += 1
  elif i == 'o':
    count += 1
  elif i == 'u':
    count += 1
  else:
    pass
print(count)
```
