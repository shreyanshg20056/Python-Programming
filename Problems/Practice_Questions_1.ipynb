Find The Last Digit of a given integer number.

```python
n = int(input("Enter a number:"))
last_digit = n % 10
print(f"Last Digit: {last_digit}")
```

Check a number is Even Or Odd without using if statement.

```python
n = int(input("Enter a no.:"))
Even = n % 2 == 0
Odd = n % 2 != 0
print("Even:",Even)
print("Odd",Odd)
```

Calculate Simple Interest.

```python
p = int(input("Enter Principal:"))
r = int(input("Enter Rate:"))
t = int(input("Enter Time:"))

SI = (p * r * t)/100
print(f"Simple Interest:{SI}")
```

Calculate Average of Three Numbers.

```python
a = int(input("Enter first number:"))
b = int(input("Enter second number:"))
c = int(input("Enter third number:"))
average = (a + b + c)/3
print("Average: ",average)
```

Find the greater between Two numbers by using Bitwise Operator.

```python
a = int(input("Enter first number:"))
b = int(input("Enter second number:"))
greater = a-((a-b) & ((a-b)>>31))
print("Graeter Value:",greater)
```

Check Whether a given Number Lies Between -10 and 20.

```python
n = int(input("Enter a no.:"))
if n >= -10 and n <= 20:
    print(f"{n} lies between -10 and 20")
else:
    print(f"{n} did not lies between -10 and 20")
```
