Calculate Factorial

```python
num = int(input("Enter a num:"))
fact = 1

for i in range(num,0,-1):
    fact = fact * i
print(f"Factorial Of A Number: {fact}")
```

Generate Fibonacci Series

```python
num =int(input("Enter a number:"))
first_num = 0
second_num = 1
next_num =  first_num + second_num
print(first_num)
print(second_num)
for i in range(num):
    next_num = next_num + i
    print(next_num)
```

Generate Palindrome

```python
num = int(input("Enter a number to check palindrome:"))
a = num // 100
b = num % 100
c = b % 10
b = b // 10
Reverse = c*100 + b*10 + a
print(f"The Given Number is:{num}")
print(f"Reverse Number = {Reverse} ")
if num == Reverse:
    print(f"{num} is a palindrome")
else:
    print(f"{num} is not a palindrome")
```

Find GCD And LCM

```python
num = int(input("Enter a number"))
num2 = int(input("Enter a number"))
for i in range(1,num+1):
    if num % i == 0:
        print(i)
N = max(num,num2)

while True:
    if N % num == 0 and N % num2 == 0:
        print(N)
        break
    N += 1
```

Count Vowels and Consonants

```python
S = input("Enter string")
vo_count = 0
co_count = 0
for i in S:
    if S == ('a','e','i','o','u'):
        vo_count += 1
    else:
        co_count = co_count + 1

print(co_count)
print(vo_count)
```
