Reverse A Word

```python
S = "Python"
for i in range(len(S) - 1,-1,-1):
  print(S[i], end = " ")
```

Count vowel in a string

```python
S = "python"
count = 0
vowel = "aeiou"

for i in S:
  if i in vowel:
    count += 1

print(count)
```

Print Fibonacci Series Upto 10 Terms

```python
n = int(input("Enter a number:"))
f = int(input("Enter first term:"))
s = int(input("Enter second term:"))
print(f)
print(s)
for i in range(0,n):
  # var1 = 0 and var2 = 1 and next = var1 + var2
  next = f + s
  f,s = s,next
  print(next)

```

Calculate Factorial Of A Number.

```python
n = int(input("Enter a no."))
fact = 1

for i in range(n,0,-1):
  fact = fact * i

print(fact)
```

Calculate Prime Number!

```python
n = int(input("Enter a number to check whether its Prime No. Or Not"))

for i in range(2,n):
  if n % i == 0:
   print(n,"Its Not Prime Number")
   break
  else:
   print(n,"Its A Prime No.")
```

Print Numbers From 10 - 15

```python
n = int(input())
while n<=15:
  print(n)
  n += 1
```

Print Cubes Of Numbers From 1 to 5 

```python
n = int(input())
while n<=5:
  cube = n*n*n
  print(cube)
  n += 1
```

Print Odd No. From 1 to 10

```python
n = int(input())
while n<=10:
  if n % 2 != 0:
    print(n)
  n += 1
```

Find The Product Of Numbers From 1-5

```python
n = 1
prod = 1
while n<=5:
  prod = prod * n
  n += 1

print(prod)
```

Calculate the power of a number

```python
Base = 3
Exp = 4
while Base <= 3:
  print(f"{Exp} Power Of a number {Base}: {pow(Base,Exp)} ")
  Base += 1
```

Calculate Perfect Square Of A Number.

```python
x = int(input())
while x <= x:
  n = x*x
  print(f"The Perfect Square Of {x}:{n}")
  x += 1
```
